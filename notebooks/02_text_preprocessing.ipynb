```python
import pandas as pd
import re
```

```python
df = pd.read_csv("../data/raw/electronics_reviews.csv")
```

```python
product_name = (
    "Logitech 915-000224  Harmony Ultimate One "
    "15-Device Universal Infrared Remote with "
    "Customizable Touch Screen Control - Black"
)

product_df = df[
    df["name"] == product_name
].copy()
```

```python
product_df.shape
```

```python
product_df["original_text"] = product_df["reviews.text"]
```

```python
product_df[
    ["original_text", "reviews.rating"]
].head()
```

```python
#Clean text
product_df["cleaned_text"] = product_df["original_text"]
```

```python
#lower text
product_df["cleaned_text"] = product_df["cleaned_text"].str.lower()
```

```python
product_df[
    ["original_text", "cleaned_text"]
].head(10)
```

```python
##Remove URLs

product_df["cleaned_text"] = product_df["cleaned_text"].apply(
    lambda text: re.sub(r"http\S+|www\S+", "", text)
)
```

```python
product_df[
    ["original_text", "cleaned_text"]
].sample(5, random_state=42)
```

```python
## punctuation with whitespace

product_df["cleaned_text"] = product_df["cleaned_text"].apply(
    lambda text: re.sub(r"[^\w\s']", " ", text)
)
```

```python
##Normalize whitespace
product_df["cleaned_text"] = product_df["cleaned_text"].apply(
    lambda text: re.sub(r"\s+", " ", text).strip()
)
```

```python
#review length

product_df["cleaned_length"] = (
    product_df["cleaned_text"]
    .str.split()
    .str.len()
)
```

```python
output_path = "../data/processed/harmony_reviews_cleaned.csv"

product_df.to_csv(
    output_path,
    index=False
)

print(f"Saved to: {output_path}")
```
