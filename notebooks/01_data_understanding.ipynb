## Data Understanding Summary

We selected the Logitech Harmony Ultimate One because it has 1,604 customer reviews,
providing sufficient data for NLP, clustering, classification, and sentiment analysis.

### Data Quality Findings

- Total reviews: 1,604
- Missing review text: 0
- Empty review text: 0
- Complete duplicate rows: 0
- Duplicate review texts: 6
- Unique reviewers: 1,500
- Missing usernames: 1
- Rating range: 1–5
- Median review length: 32 words
- Mean review length: approximately 69 words
- Maximum review length: 1,731 words

### Key observations

The review lengths are highly variable. Most reviews are short, while a small
number are very long.

The dataset is also skewed toward higher ratings, with substantially more
4- and 5-star reviews than lower ratings.

Duplicate review text was found in only six records. These were retained because
identical text does not necessarily mean that the underlying review records are
duplicates.

For downstream NLP tasks, the original review text will be preserved while a
separate cleaned text representation will be created for traditional NLP methods.
